```python
import pandas as pd
from pathlib import Path

# ============================================================
# PATH
# ============================================================

OUTPUT_DIR = Path(
    r"..\Output"
)

INPUT_FILE = OUTPUT_DIR / "03_Hasil_Network_OSRM_FINAL.csv"

OUTPUT_MALUKU = OUTPUT_DIR / "15_Hasil_Network_Maluku.csv"
OUTPUT_MALUKU_UTARA = OUTPUT_DIR / "15_Hasil_Network_Maluku_Utara.csv"


# ============================================================
# 1. MEMBACA DATA
# ============================================================

print("Membaca file network hasil OSRM...")

df = pd.read_csv(
    INPUT_FILE,
    dtype={"KDDESA": str}
)

print(f"Total data: {len(df)}")


# ============================================================
# 2. CEK PROVINSI
# ============================================================

print("\nProvinsi yang tersedia:")
print(df["NMPROP"].value_counts())


# ============================================================
# 3. PISAHKAN DATA
# ============================================================

maluku = df[
    df["NMPROP"]
    .astype(str)
    .str.strip()
    .str.upper()
    == "MALUKU"
].copy()

maluku_utara = df[
    df["NMPROP"]
    .astype(str)
    .str.strip()
    .str.upper()
    == "MALUKU UTARA"
].copy()


# ============================================================
# 4. SIMPAN
# ============================================================

maluku.to_csv(
    OUTPUT_MALUKU,
    index=False,
    encoding="utf-8-sig"
)

maluku_utara.to_csv(
    OUTPUT_MALUKU_UTARA,
    index=False,
    encoding="utf-8-sig"
)


# ============================================================
# 5. RINGKASAN
# ============================================================

print("\n" + "=" * 60)
print("SELESAI")
print("=" * 60)

print(f"Maluku       : {len(maluku)} desa")
print(f"Maluku Utara : {len(maluku_utara)} desa")

print("\nOutput:")
print(OUTPUT_MALUKU)
print(OUTPUT_MALUKU_UTARA)
```
