```python
from pathlib import Path
import pandas as pd


# ============================================================
# PATH
# ============================================================

PROJECT_ROOT = Path.cwd().parent  # run from the Script/ folder

OUTPUT_DIR = PROJECT_ROOT / "Output"


INPUT_FILE = OUTPUT_DIR / (
    "01_Hasil_Euclidean_Desa_FKTP.csv"
)


# ============================================================
# OUTPUT FILES
# ============================================================

MALUKU_OUT = OUTPUT_DIR / (
    "01_Hasil_Euclidean_Maluku.csv"
)


MALUKU_UTARA_OUT = OUTPUT_DIR / (
    "01_Hasil_Euclidean_Maluku_Utara.csv"
)


# ============================================================
# MAIN
# ============================================================

def main():

    # --------------------------------------------------------
    # 1. BACA DATA
    # --------------------------------------------------------

    df = pd.read_csv(

        INPUT_FILE

    )


    print(

        f"Total data awal: {len(df)}"

    )


    # --------------------------------------------------------
    # 2. CEK PROVINSI
    # --------------------------------------------------------

    print()

    print(

        "Provinsi yang tersedia:"

    )


    print(

        df["NMPROP"]

        .value_counts()

        .to_string()

    )


    # --------------------------------------------------------
    # 3. PISAHKAN MALUKU
    # --------------------------------------------------------

    maluku = df[

        df["NMPROP"]

        .astype(str)

        .str.strip()

        .str.upper()

        == "MALUKU"

    ].copy()


    # --------------------------------------------------------
    # 4. PISAHKAN MALUKU UTARA
    # --------------------------------------------------------

    maluku_utara = df[

        df["NMPROP"]

        .astype(str)

        .str.strip()

        .str.upper()

        == "MALUKU UTARA"

    ].copy()


    # --------------------------------------------------------
    # 5. SIMPAN FILE
    # --------------------------------------------------------

    maluku.to_csv(

        MALUKU_OUT,

        index=False,

        encoding="utf-8"

    )


    maluku_utara.to_csv(

        MALUKU_UTARA_OUT,

        index=False,

        encoding="utf-8"

    )


    # --------------------------------------------------------
    # 6. HASIL
    # --------------------------------------------------------

    print()

    print("=" * 60)

    print(

        "HASIL PEMISAHAN"

    )

    print("=" * 60)


    print(

        f"Maluku       : {len(maluku)} data"

    )


    print(

        f"Maluku Utara : {len(maluku_utara)} data"

    )


    print()

    print(

        "File Maluku tersimpan di:"

    )


    print(

        MALUKU_OUT

    )


    print()

    print(

        "File Maluku Utara tersimpan di:"

    )


    print(

        MALUKU_UTARA_OUT

    )


# ============================================================
# RUN
# ============================================================

if __name__ == "__main__":

    main()
```
